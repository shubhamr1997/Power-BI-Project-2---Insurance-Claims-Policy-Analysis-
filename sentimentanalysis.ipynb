```python
# Customer Feedback Sentiment Analysis
#
# This notebook reads the Excel file at the specified path and computes sentiment scores
# for each customer feedback entry.
#
# It adds:
# - sentiment_score: normalized score from 0 to 1
```

```python
import pandas as pd
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer

# Install/download required NLTK data
nltk.download('vader_lexicon')

# Path to your feedback file
file_path = r"C:\Users\priya\Desktop\All Projects\Power BI\Power BI Project 2\Data\Insurance+Customer+Feedback.xlsx"

# Load the data
# If your feedback is on a different sheet, set sheet_name='SheetName'
df = pd.read_excel(file_path)

# Locate the feedback column
feedback_candidates = [
    col for col in df.columns
    if isinstance(col, str) and col.strip().lower() in {
        'feedback', 'review', 'comments', 'customer feedback', 'customer comments', 'message'
    }
]

if feedback_candidates:
    feedback_col = feedback_candidates[0]
else:
    text_columns = [col for col in df.columns if df[col].dtype == object]
    if len(text_columns) == 0:
        raise ValueError('No text column found for sentiment analysis. Please check the Excel file columns.')
    feedback_col = text_columns[0]

print('Using feedback column:', feedback_col)

# Create sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Compute normalized sentiment score from 0 to 1
score_series = df[feedback_col].fillna('').astype(str).apply(
    lambda text: (sia.polarity_scores(text)['compound'] + 1) / 2
)
df['sentiment_score'] = score_series

# Preview results
print(df[[feedback_col, 'sentiment_score']].head())

# Save results back to Excel
output_path = r"C:\Users\priya\Desktop\All Projects\Power BI\Power BI Project 2\Data\Insurance+Customer+Feedback_sentiment.xlsx"
df.to_excel(output_path, index=False)
print('Saved sentiment output to:', output_path)
```
